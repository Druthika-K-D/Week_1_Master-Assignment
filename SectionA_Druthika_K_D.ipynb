Q7. What is the difference between `list.append(x)`, `list.extend(x)`, and `list.insert(i, x)`? For each,
show what happens when `x = [4, 5]` is added to `a = [1, 2, 3]`.

7.ans)list.append(x) adds the entire list x as a single element at the end of list 'a'.
output:[1, 2, 3, [4, 5]]

list.extend(x) adds each element of x separately at the end one by one.

output:[1, 2, 3, 4, 5]

list.insert(i, x) adds the entire list x at the index i.

output:[1, [4, 5], 2, 3]
