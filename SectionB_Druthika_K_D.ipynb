```python
#Q14. Predict the output of each print statement. Then state which lines will raise an error (if any).
print("10" + "20")
print(10 + 20)
print("10" * 3)
print(10 * "3")
print("10" + 20)
print(int("10") + 20)
```

print("10"+"20") gives 1020 because both are enclosed with double quotes making them string values,so it gets concatenated instead of added.

print(10+20) gives 30 because addition operation is performed,as both vales are integers.

print("10"*3) gives 101010 because 10 is string datatype here and 3 is integer,* operator is used which is used to replicate 10 ,3 times.

print(10*"3") gives 3333333333 as here 3 is replicated 10 times.

print("10"+20) gives error because we cannot perform addition on string with integer.

print(int("10")+20) gives 30 because the string type is converted to int type.

```python
#Q15. Predict the output. Then explain Python's short-circuit evaluation in one line.
print(0 or 5)
print(5 or 0)
print(None or "hello")
print(0 and 5)
print(5 and 10)
print([] or "default")
print([0] or "default") # careful!
```

here true values are returned,
0 or 5 - 5 beacuse 0 is false,

5 or 0 - 5 because true,

None or "hello" - hello because None is false

0 and 5 - 0 because it is false(both values must be true for and)

5 and 10 - 10 because both are true so it returns second value

[0] or "default"  in case of [0] ,it is not an empty list so it returns true which will print [0].
